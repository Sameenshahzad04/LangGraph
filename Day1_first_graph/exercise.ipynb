```python
from typing import Dict, TypedDict
from langgraph.graph import StateGraph
from  IPython.display import display,Image

```

```python
class AgentState(TypedDict):
    name: str
    messages: str
    
# defining a node:a step in the graph that represents a specific task or action.in general python function can be a node in the graph, but we can also define custom nodes that encapsulate specific logic or behavior.

def Compliment(state: AgentState) -> AgentState:

    """Simple node that adds a compliment message to the state"""

    state['messages'] = state['name']+" you`re doing an amazing job!."
    
    return state
```

```python
#create a graph
graph = StateGraph(AgentState)
graph.add_node("Compliment Agent", Compliment)
graph.set_entry_point("Compliment Agent")
graph.set_finish_point("Compliment Agent")

app=graph.compile()
```

```python
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
result=app.invoke({"name": "bob"})
result['messages']
```
